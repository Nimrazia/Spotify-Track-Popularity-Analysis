Exploratory Data Analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("tracks2026.csv")
print("Shape:", df.shape)
```

```python
display(df.describe())

display(df.describe(include="object"))
```

```python
#Missing Values
missing = df.isnull().sum().sort_values(ascending=False)
missing_percent = (df.isnull().mean() * 100).sort_values(ascending=False)

missing_df = pd.DataFrame({
    "missing_count": missing,
    "missing_percent": missing_percent
})

display(missing_df[missing_df["missing_count"] > 0])
```

```python
print("Duplicate rows:", df.duplicated().sum())

if "track_id" in df.columns:
    print("Duplicate track_id values:", df["track_id"].duplicated().sum())
    display(df[df["track_id"].duplicated(keep=False)].sort_values("track_id").head(20))
```

```python

df = df.drop_duplicates()

print("New shape after removing duplicates:", df.shape)
```

```python
display(df.sort_values("loudness", ascending=False).head(10))
```

```python
# replace unrealistic loudness values with NaN
df.loc[df["loudness"] > 0, "loudness"] = np.nan

print("Updated loudness summary:")
display(df["loudness"].describe())
```

```python
df.hist(figsize=(10, 14), bins=30)
plt.tight_layout()
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include=["object", "bool"]).columns.tolist()

print("Numeric columns:")
print(numeric_cols)

print("\nCategorical columns:")
print(categorical_cols)
```

```python
df[numeric_cols].hist(figsize=(18, 14), bins=30, edgecolor="black")
plt.suptitle("Distributions of Numeric Features", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(18, 14))

n_cols = 3
n_rows = int(np.ceil(len(numeric_cols) / n_cols))

for i, col in enumerate(numeric_cols, 1):
    plt.subplot(n_rows, n_cols, i)
    sns.boxplot(x=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
#Popularity Analysis
plt.figure(figsize=(8, 5))
sns.histplot(df["popularity"], bins=30, kde=True)
plt.title("Distribution of Popularity")
plt.xlabel("Popularity")
plt.ylabel("Count")
plt.show()

print("Popularity summary:")
display(df["popularity"].describe())
```

```python
median_popularity = df["popularity"].median()
print("Median popularity:", median_popularity)
```

```python
plt.figure(figsize=(8, 3))
sns.boxplot(x=df["popularity"])
plt.title("Boxplot of Popularity")
plt.show()
```

```python
if "track_genre" in df.columns:
    plt.figure(figsize=(10, 5))
    genre_counts = df["track_genre"].value_counts()
    sns.barplot(x=genre_counts.index, y=genre_counts.values)
    plt.title("Track Count by Genre")
    plt.xlabel("Genre")
    plt.ylabel("Count")
    plt.xticks(rotation=30)
    plt.show()

    display(genre_counts)
```

```python
if "explicit" in df.columns:
    plt.figure(figsize=(5, 4))
    sns.countplot(x=df["explicit"])
    plt.title("Distribution of Explicit Content")
    plt.xlabel("Explicit")
    plt.ylabel("Count")
    plt.show()

    display(df["explicit"].value_counts())
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(12, 8))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.show()

# correlation with popularity
pop_corr = corr["popularity"].sort_values(ascending=False)
display(pop_corr)
```

```python
key_features = ["danceability", "energy", "acousticness", "valence", "tempo"]

available_features = [col for col in key_features if col in df.columns]

plt.figure(figsize=(15, 10))

for i, col in enumerate(available_features, 1):
    plt.subplot((len(available_features) + 1) // 2, 2, i)
    sns.scatterplot(data=df, x=col, y="popularity", alpha=0.5)
    plt.title(f"{col} vs Popularity")

plt.tight_layout()
plt.show()
```

```python
def count_outliers_iqr(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return ((series < lower) | (series > upper)).sum()

outlier_counts = pd.DataFrame({
    "feature": numeric_cols,
    "outlier_count": [count_outliers_iqr(df[col].dropna()) for col in numeric_cols]
}).sort_values("outlier_count", ascending=False)

display(outlier_counts)
```
